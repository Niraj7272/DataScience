### 91. Implement one-hot encoding using NumPy only.

### 96. Simulate rolling dice 1000 times and calculate probabilities. 

```python
# Simulate rolling a die 1000 times
rolls = np.random.randint(1, 7, 1000)

# Count occurrences of each face
counts = np.bincount(rolls)[1:]   # Ignore index 0

# Calculate probabilities
probabilities = counts / len(rolls)

print("Counts:", counts)
print("Probabilities:", probabilities)

# Display each face and its probability
for i, prob in enumerate(probabilities, start=1):
    print(f"Face {i}: {prob:.3f}")
```

### 98. Implement simple linear regression using NumPy only. 

```python
# Sample data
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# Calculate slope (m) and intercept (b)
m = np.sum((x - np.mean(x)) * (y - np.mean(y))) / np.sum((x - np.mean(x)) ** 2)
b = np.mean(y) - m * np.mean(x)

# Predicted values
y_pred = m * x + b

print("Slope (m):", m)
print("Intercept (b):", b)
print("Predicted values:", y_pred)
```

### 100. Build a mini student marks analysis system using NumPy (average, topper, pass/fail, subject-wise         statistics). 

```python
# Student names
students = np.array(["Alice", "Bob", "Charlie", "David", "Emma"])

# Marks (Rows = Students, Columns = Subjects)
# Subjects: Math, Science, English
marks = np.array([
    [85, 78, 92],
    [65, 70, 60],
    [95, 88, 91],
    [45, 55, 50],
    [75, 82, 79]
])

subjects = np.array(["Math", "Science", "English"])

# ----------------------------
# Student-wise Analysis
# ----------------------------

# Total marks
total = np.sum(marks, axis=1)

# Average marks
average = np.mean(marks, axis=1)

print("Student Report")
print("-" * 40)

for i in range(len(students)):
    status = "Pass" if np.all(marks[i] >= 40) else "Fail"

    print(f"{students[i]}")
    print(f"Total   : {total[i]}")
    print(f"Average : {average[i]:.2f}")
    print(f"Status  : {status}")
    print()

# ----------------------------
# Class Statistics
# ----------------------------

class_average = np.mean(total)

topper_index = np.argmax(total)

print("-" * 40)
print("Class Statistics")
print("-" * 40)

print("Class Average:", round(class_average, 2))
print("Topper:", students[topper_index])
print("Topper Marks:", total[topper_index])

# ----------------------------
# Subject-wise Statistics
# ----------------------------

print("\nSubject-wise Statistics")
print("-" * 40)

subject_avg = np.mean(marks, axis=0)
subject_max = np.max(marks, axis=0)
subject_min = np.min(marks, axis=0)

for i in range(len(subjects)):
    print(subjects[i])
    print("Average :", round(subject_avg[i], 2))
    print("Highest :", subject_max[i])
    print("Lowest  :", subject_min[i])
    print()
```
