### 71. Find determinant of a matrix. 

```python
matrix = np.array([
    [1, 2],
    [3, 4]
])

det = np.linalg.det(matrix)

print(det)
```

### 72. Find inverse of a matrix. 

```python
matrix = np.array([
    [1, 2],
    [3, 4]
])
inverse = np.linalg.inv(matrix)
print(inverse)
```

### 73. Find eigenvalues of a matrix. 

```python
matrix = np.array([
    [4, 2],
    [1, 3]
])
result = np.linalg.eigvals(matrix)
print(result)
```

### 74. Find eigenvectors of a matrix. 

```python
matrix = np.array([
    [4, 2],
    [1, 3]
])


eigenvectors = np.linalg.eig(matrix)

print("Eigenvectors:")
print(eigenvectors)
```
