```python
category_subscribers_mean = df.groupby('category')['subscribers'].mean()
```

```python
category_subscribers_mean
```

```python
category_subscribers_mean.sort_values(ascending=False, inplace=True)
```

```python
category_subscribers_mean
```

```python
category_subscribers_mean.plot(kind='bar')
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.xticks(rotation = 90)
plt.show()
```

```python
category_subscribers_mean.head().plot(kind='bar')
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.xticks(rotation = 90)
plt.show()
```

```python
category_subscribers_mean.tail().plot(kind='bar')
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.xticks(rotation = 90)
plt.show()
```

```python
sns.boxplot(x='channel_type', y='highest_monthly_earnings', data=df)
plt.title('Highest Monthly Earnings by Channel Type')
plt.xlabel('Channel Type')
plt.ylabel('Highest Monthly Earnings')  
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby(['category','Country'])['subscribers'].sum().unstack().plot(kind='bar', stacked=True, figsize=(12,8))    
plt.title('Total Subscribers by Category and Country')
plt.xlabel('Category')
plt.ylabel('Total Subscribers')
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.scatter(df['Gross tertiary education enrollment (%)'], df['subscribers'])
plt.title('Subscribers vs Gross Tertiary Education Enrollment')
plt.xlabel('Gross Tertiary Education Enrollment (%)')
plt.ylabel('Subscribers')
plt.show()
```

```python
df.groupby('channel_type')['Unemployment rate'].mean().head().sort_values(ascending=False).plot(kind='bar')
plt.title('Average Unemployment Rate by Channel Type')
plt.xlabel('Channel Type')
plt.ylabel('Average Unemployment Rate')
plt.xticks(rotation=90)
plt.show()
```

```python
df.groupby(['created_year', 'channel_type'])['subscribers'].mean().unstack().tail().plot(kind='bar', width=1)
plt.title('Average Subscribers by Created Year and Channel Type')
plt.xlabel('Created Year')
plt.ylabel('Average Subscribers')
plt.xticks(rotation=90)
plt.legend(title='Channel Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
df.groupby(['created_year', 'created_month'])['subscribers'].sum().unstack().plot(kind='area', stacked=True, figsize=(12,8))
plt.title('Total Subscribers by Created Year and Month')
plt.xlabel('Created Year')
plt.ylabel('Total Subscribers')
plt.legend(title='Created Month', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=90)
plt.show()
```

```python
df.groupby(['created_year', 'Country'])['highest_yearly_earnings'].mean().unstack().head().plot(kind='line', marker='o' )
plt.title('Trend of Highest Yearly Earnings by Created Year and Country')
plt.xlabel('Created Year')
plt.ylabel('Average Highest Yearly Earnings')
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=90)
plt.show()
```
