```python
category_subscribers_mean = df.groupby('category')['subscribers'].mean()
```

```python
category_subscribers_mean
```

```python
category_subscribers_mean.sort_values(ascending=False, inplace=True)
```

```python
category_subscribers_mean
```

```python
category_subscribers_mean.plot(kind='bar')
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.xticks(rotation = 90)
plt.show()
```

```python
category_subscribers_mean.head().plot(kind='bar')
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.xticks(rotation = 90)
plt.show()
```

```python
category_subscribers_mean.tail().plot(kind='bar')
plt.title('Average Subscribers by Category')
plt.xlabel('Category')
plt.ylabel('Average Subscribers')
plt.xticks(rotation = 90)
plt.show()
```

```python
channel_type_counts = df['channel_type'].value_counts()
channel_type_counts
```

```python
channel_type_counts.plot(kind='pie')
plt.title('Channel Type Distribution')
plt.show()
```

```python
channel_type_counts.head().plot(kind='pie', autopct='%1.1f%%')
plt.title('Channel Type Distribution')
plt.show()
```

```python
channel_type_counts.tail().plot(kind='pie', autopct='%1.1f%%')
plt.title('Channel Type Distribution')
plt.show()
```

```python
plt.scatter(df['subscribers'], df['video views'])
plt.title('Subscribers vs Views')
plt.xlabel('Subscribers')
plt.ylabel('Views')
plt.show()
```

```python
country_subscribers_total = df.groupby('Country')['subscribers'].sum().sort_values(ascending=False)
```

```python
country_subscribers_total
```

```python
country_subscribers_total.head(10).plot(kind='bar')
plt.title('Total Subscribers by Country')
plt.xlabel('Country')
plt.ylabel('Total Subscribers')
plt.xticks(rotation= 90)
plt.show()
```
