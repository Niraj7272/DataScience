```python
plt.figure(figsize=(15,8))

x = [1,2,3,4]

sales = [10, 20, 25, 30]
profit = [5, 10, 15, 20]

plt.plot(x,sales, label='Sales')
plt.plot(x,profit, label='Profit')
plt.legend()
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 120, 90, 140]

plt.plot(months,sales, label='Sales')
plt.title('Monthly Sales Report', fontsize=18,color='red', fontweight='bold')
plt.xlabel('Months', color='green', fontsize=12)
plt.ylabel('Sales', color='green', fontsize=12)
plt.legend(loc='upper left')
plt.show()
```

```python
fig , ax = plt.subplots(2,1)

ax[0].plot([1,2,3],[4,5,6])
ax[0].set_title("Line Plot")

ax[1].bar(
    ['A','B','C'],
    [5,7,3]
)
ax[1].set_title("Bar chat")

plt.tight_layout()
plt.subplots_adjust(hspace=0.5)
plt.show()
```

```python
fig, ax = plt.subplots(1,2)
ax[0].plot([1,2,3],[2,4,6])
ax[1].scatter([1,2,3],[3,6,9])
plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr']
sales = [100, 120, 90, 140]
profit = [20, 25, 18, 30]

fig, ax = plt.subplots(
    2,
    1,
    figsize = (8,6)
)

ax[0].plot(
    months,
    sales,
    marker ='o',
    linestyle = '-'
)
ax[0].set_title("Monthly Sales")
ax[0].grid()

ax[1].bar(
    months,
    profit,
    color ='green'
)
ax[1].set_title("Monthly Profit")
plt.tight_layout()
plt.savefig('business_report.png', dpi=600)
plt.show()
```
