## 20. Classify Movies Based on Ratings [Excellent,Good,Average and Bad]

```python
df.columns
```

```python
def rating(rating):
    if rating >= 7.0:
        return 'Excellent'
    elif rating >= 6.0:
        return 'Good'
    elif rating >= 5.0:
        return 'Average'
    else:
        return 'Bad'
```

```python
rating(8)
```

```python
rating(6.5)
```

```python
df['rating_category'] = df['Rating'].apply(rating)
```

## 22. Find Unique Values From Genre

```python
df.columns
```

```python
df['Genre']
```

```python
'Action,Adventure,Sci-Fi'.split(',')
```

```python
list1 = []
for value in df['Genre']:
    list1.append(value.split(','))
```

```python
list1
```

```python
one_dimension = []
for i in list1:
    for value in i:
        one_dimension.append(value)
```
