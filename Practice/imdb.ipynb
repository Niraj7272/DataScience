## 10. In Which Year There Was The Highest Voting?

## 13. Display Top 10 Lengthy Movies Title

```python
df.columns
```

```python
df.nlargest(10,'Runtime (Minutes)')['Title']
```

```python
le = df.nlargest(10, 'Runtime (Minutes)')[['Title', 'Runtime (Minutes)']].set_index('Title')
le
```

```python
sns.barplot(data=le[:5], x =le['Runtime (Minutes)'].head(),y=le.index[:5])
plt.title('Top 5 Lengthy Movies')
plt.show()
```
