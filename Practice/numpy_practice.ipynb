start : Starting value of the range.

end : Ending value of the range(inclusive).

num (optional) : Number of values to create

### Array Indexing
In NumPy arrays, just like in Python lists, each element has an index number that starts from 0

In NumPy arrays, negative indexing lets you access elements from the end. For example:

-1 gives the last element.

-2 gives the second-to-last element.

If the array is [1, 2, 3, 4, 5]:

array[-1] is 5

array[-2] is 4.

```python
arr = np.array([1,2,3,4,5])
arr
```

```python
arr[3]
```

```python
arr[5]
```

```python
print(arr[0])
print(arr[4])
print(arr[5])
```

```python
print(arr[0])
print(arr[3])
try:
    print(arr[5])
except Exception as e:
    print(e)
```
