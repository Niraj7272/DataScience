### Negative slicing in NumPy allows you to access elements from the end of an array using negative indices.

```python
# Negative slicing

# create the 1-D numpy array
arr = np.array([1,2,3,4,5,6])

# slice the last 3 elements of the array
slice_arr = arr[-3:]
print(slice_arr)
```

```python
arr = np.array([1,2,3,4,5,6])

# slice the array in reverse order
slice_arr = arr[-1::-1]
print(slice_arr)
```

```python
# create a numpy array
numbers = np.array([2,4,6,8,10,12,14])

# create reverse array
reverse_numbers = numbers[::-1]
print(reverse_numbers)
```

### Array Attributes

Attributes and Their Meanings:

ndim: Number of dimensions (e.g., 1 for a 1D array).

shape: Size of the array in each dimension (e.g., length for 1D).

size: Total number of elements in the array.

dtype: Type of data stored in the array.

itemsize: Size of each element in bytes.

nbytes: Total memory used by the array (itemsize × size).

```python
# create a numpy array
numbers = np.array([2,4,6,8,10,12,14])

# give the array dimension
print('Number of dimension:', numbers.ndim)

# get shape of the array
print('Shape:',numbers.shape)
```

```python
number1 = np.array([2,4,6,8,10,'a'])
print(number1)
print('Data type of first array:',number1.dtype)

# create an array of int type
int_numbers = np.array([2,4,6,8,10,12,13])

# create an array of float type
float_numbers = np.array([3.4,5.6,7.8,9.1])


# check data type of numbers
print('Data type of second array:',int_numbers.dtype)

# check data type of float array
print('Data type of third array:',float_numbers.dtype)
```

```python
# we use size attribute to find the total number of elements in the array
first = np.array([1,3,4,6,8])
second = np.array([2,4,6,8,1,10,13,12])

print('size of the first array:',first.size)
print('size of the second array:',second.size)
```

```python
# We use itemsize attribute to find the size(in bytes) of each element in the array.
# create a default 1-D array of integers
array1 = np.array([6,7,8,9,10,11])

# create a 1-D array of 16-bit integers
array2 = np.array([6,7,8,9,10,11], dtype=np.int16)

# create a 1-D array of floats
array3 = np.array([1.2,2.4,5.6,7.8], dtype= np.float64)

print(array1)
print(array2)
print(array3)

print(array1.itemsize)
print(array2.itemsize)
print(array3.itemsize)
```

```python
# nbytes finds the total number of bytes consumed by elements in the array.
print(array1.nbytes)
print(array2.nbytes)
print(array3.nbytes)
```

```python
array4 = np.array([1,2,3,4,5],dtype= np.float32)
print(array4)
print(array4.itemsize)
print(array4.nbytes)
```
