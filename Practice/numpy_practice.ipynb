np.arange() function in NumPy works like Python's range() to create sequences of numbers. Its syntax is: np.arange(start,end,step)

```python
list(range(5))
```

```python
list(range(5,15))
```

```python
list(range(5,15,2))
```

```python
list(range(5,15,2.2))
```

```python
# comparison of range and arange
a = np.arange(5)
a
```

```python
b = np.arange(5,20)
b
```

```python
c = np.arange(3,40,4)
c
```

```python
d = np.arange(5,15,2.2)
d
```

np.linspace() function creates an array with evenly spaced numbers within a range. It's syntax is : np.linspace(start,end,num)

start : Starting value of the range.

end : Ending value of the range(inclusive).

num (optional) : Number of values to create

```python
f = np.linspace(5,500,67)
f
```

```python
12.5 - 5, 20-12.5,27.5-20
```

```python
np.linspace(1,100,10)
```
