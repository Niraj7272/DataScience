- word Tokenization

### Stemming

Types of Stemming
- Porter Stemmer
- Regexp Stemmer
- Snowball Stemmer

```python
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalized']
words
```

```python
# Porter Stemmer
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word + '---->' + stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem('sitting')
```

```python
# RegexpStemmer
```

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
# Snowball Stemmer
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
paragraph = """
Nepal is a beautiful country located in South Asia.
It is famous for its majestic mountains, including Mount Everest, the highest mountain in the world.
Nepal has a rich culture with many festivals, traditions, and historic temples.
Kathmandu is the capital city and an important center of culture and history.
The country is home to many different ethnic groups, languages, and religions.
Nepal is also known for its natural beauty, including mountains, rivers, forests, and national parks.
I am proud of Nepal because of its unique culture, friendly people, and beautiful landscapes.
"""
```

```python
from nltk.stem import porter
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
stopwords.words('arabic')
```

```python
stopwords.words('nepali')
```

```python
stemmer = PorterStemmer()
```

```python
paragraph
```

```python
sentences = nltk.sent_tokenize(paragraph)
sentences
```

```python
type(sentences)
```

```python
len(sentences)
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer('english')
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [snowballstemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word.lower(), pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
```
