## BOW (Bag of words)

```python
import pandas as pd
```

```python
messages = pd.read_csv('SMSSpamCollection', sep='\t', names=['label','message'])
messages
```

```python
# Data Cleaning and Preprocessing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
messages.shape
```

```python
len(messages)
```

```python
corpus = []
for i in range(len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
# Create Bag of Words
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=100, binary=True)
cv
```

```python
X = cv.fit_transform(corpus).toarray()
X
```

```python
import numpy as np
np.set_printoptions(
    edgeitems=30, 
    linewidth=10000, 
    formatter=dict(float=lambda x: "{0:0.3g}".format(x))
)
```

```python
X
```

```python
cv.vocabulary_
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=100, binary=True, ngram_range=(2,3))
X = cv.fit_transform(corpus).toarray()
```

```python
X
```

```python
cv.vocabulary_
```

## TFIDF

```python
messages = pd.read_csv('SMSSpamCollection', sep='\t', names=['label','message'])
messages
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

wordlemmatizer = WordNetLemmatizer()
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [wordlemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=100)
X = tfidf.fit_transform(corpus).toarray()
X
```

```python
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float= lambda x: "{0:0.3g}".format(x)))
```

```python
X
```

```python
tfidf = TfidfVectorizer(max_features=100, ngram_range=(2,3))
X = tfidf.fit_transform(corpus).toarray()
```

```python
tfidf.vocabulary_
```
