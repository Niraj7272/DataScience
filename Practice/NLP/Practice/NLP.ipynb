### Tokenization

```python
corpus = """Hello Welcome to my NLP Session.
You can watch recordings for revision. We're learning tokenization.
We are exploring tokenization."""
```

```python
corpus
```

```python
type(corpus)
```

```python
print(corpus)
```

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus)
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
len(documents)
```

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
words = word_tokenize(corpus)
```

```python
type(words)
```

```python
len(words)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
len(wordpunct_tokenize(corpus))
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
len(tokenizer.tokenize(corpus))
```

### Stemming

Types of Stemming
- Porter Stemmer
- Regexp Stemmer
- Snowball Stemmer

```python
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalized']
words
```

```python
# Porter Stemmer
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word + '---->' + stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem('sitting')
```

```python
# RegexpStemmer
```

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
# Snowball Stemmer
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
words
```

```python
for word in words:
    print(word + '--->' + snowballstemmer.stem(word))
```

```python
stemming.stem('fairly'), stemming.stem('sportingly')
```

```python
snowballstemmer.stem('fairly'), snowballstemmer.stem('sportingly')
```

```python
snowballstemmer.stem('goes')
```

```python
stemming.stem('goes')
```

```python
stemming.stem('congratulations'), snowballstemmer.stem('congratulations')
```

### Lemmatization

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS Noun --> n
POS Verb --> v
POS Adjective --> a
POS Adverb --> r
'''

lemmatizer.lemmatize('going', pos='v')
```

```python
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalized']
```

```python
words
```

```python
for word in words:
    print(word + '--->' + lemmatizer.lemmatize(word, pos='v'))
```
