XGBoost stands for Extreme Gradient Boosting. It's a machine learning algorithm used mostly for classification and regression tasks

Imagine you're a student trying to guess the answers on a test. You make a few guesses and get some wrong.

Next time, you focus on the questions you got wrong.

You try again, improving your guesses by learning from past mistakes.

After doing this many times, your final answers become really good.

XGBoost works just like this — it builds many small decision trees one after another, each time trying to fix the mistakes of the previous ones.

How It Works (Steps):

Make a guess (first tree).

Check the errors.

Make a new tree to fix those errors.

Repeat this process many times.

Add up all the trees' guesses to get a final strong prediction.

Why It's Good: Fast and efficient.

Handles missing values well.

Can handle large datasets.

Often gives very accurate prediction

### Imputing Null Values

1. Impute Median for 'Age' column
2. Impute Mode for 'TypeofContact' column
3. Impute Median for 'DurationOfPitch' column
4. Impute Mode for 'NumberofFollowups' column as it is discrete  feature
5. Impute Mode for 'PreferredPropertyStar' column
6. Impute Median for 'NumberOfTrips' column
7. Impute Mode for 'NumberOfChildrenVisiting' column
8. Impute Median for 'MonthlyIncome' column

```python
# Age
df.Age.fillna(df.Age.mean(), inplace=True)

# TypeofContact
df.TypeofContact.fillna(df.TypeofContact.mode()[0], inplace=True)

# DurationOfPitch
df.DurationOfPitch.fillna(df.DurationOfPitch.median(), inplace=True)

# NumberofFollowups
df.NumberOfFollowups.fillna(df.NumberOfFollowups.mode()[0], inplace=True)

#PreferredPropertyStar
df.PreferredPropertyStar.fillna(df.PreferredPropertyStar.mode()[0], inplace=True)

# NumberOfTrips
df.NumberOfTrips.fillna(df.NumberOfTrips.median(), inplace=True)

# NumberOfChildrenVisiting
df.NumberOfChildrenVisiting.fillna(df.NumberOfChildrenVisiting.mode()[0], inplace=True)

# MonthlyIncome
df.MonthlyIncome.fillna(df.MonthlyIncome.median(), inplace=True)
```
