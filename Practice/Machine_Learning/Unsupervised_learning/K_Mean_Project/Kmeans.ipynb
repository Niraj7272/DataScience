```python
df.columns
```

```python
corr = df[['Income','Age','Recency','Total_Spending','NumWebPurchases','NumStorePurchases']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
pivot_income = df.pivot_table(
    index='Education',
    values='Income',
    columns='Marital_Status',
    aggfunc='mean'
)
pivot_income
```

```python
sns.heatmap(pivot_income, annot=True, cmap='YlGnBu', fmt='.0f')
plt.title('Income Distribution by Education and Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Education Level')
plt.show()
```

```python
group1 = df.groupby('Education')['Income'].mean().sort_values(ascending=False)
group1
```

```python
group1.plot(kind='bar', color='skyblue')
plt.title('Average Income by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Average Income')
plt.show()
```

```python
group2 = df.groupby('Education')['Total_Spending'].mean().sort_values(ascending=False)
group2
```

```python
group2.plot(kind='bar', color='salmon')
plt.title("Average Total Spending by Education Level")
plt.xlabel('Education Level')
plt.ylabel('Average Total Spending')
plt.xticks(rotation=45)
plt.show()
```
